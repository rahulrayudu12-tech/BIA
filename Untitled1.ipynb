```python
import pandas as pd

data = {
    "Review": [
        "The product is excellent",
        "I really loved this product",
        "Amazing quality and great service",
        "Very happy with my purchase",
        "The product works perfectly",
        "Good quality for the price",
        "The product is terrible",
        "I hated this product",
        "Very poor quality",
        "The product stopped working",
        "Waste of money",
        "Extremely disappointed",
        "It is an average product",
        "The product is okay",
        "Nothing special about this product",
        "The quality is acceptable",
        "It works as expected",
        "The product is neither good nor bad"
    ],

    "Sentiment": [
        "Positive", "Positive", "Positive", "Positive",
        "Positive", "Positive",
        "Negative", "Negative", "Negative", "Negative",
        "Negative", "Negative",
        "Neutral", "Neutral", "Neutral", "Neutral",
        "Neutral", "Neutral"
    ]
}

df = pd.DataFrame(data)

df.to_csv("customer_reviews.csv", index=False)

print("Dataset created successfully!")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv("customer_reviews.csv")

print(df.head())
```

```python
print("Number of records:", len(df))
```

```python
import re

def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    return text

df["Review"] = df["Review"].apply(preprocess_text)

print(df.head())
```

```python
X = df["Review"]
y = df["Sentiment"]
```

```python
vectorizer = CountVectorizer(stop_words='english')

X_vectorized = vectorizer.fit_transform(X)

print(X_vectorized.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_vectorized,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)
```

```python
model = MultinomialNB()

model.fit(X_train, y_train)

print("Naive Bayes model trained successfully!")
y_pred = model.predict(X_test)

print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Accuracy percentage:", accuracy * 100, "%")
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
plt.figure(figsize=(6, 4))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Confusion Matrix")

plt.show()
print(classification_report(y_test, y_pred))
```

```python
new_reviews = [
    "This product is amazing",
    "Very poor product",
    "The product is okay"
]

new_reviews_clean = [preprocess_text(review) for review in new_reviews]

new_reviews_vectorized = vectorizer.transform(new_reviews_clean)

predictions = model.predict(new_reviews_vectorized)

for review, sentiment in zip(new_reviews, predictions):
    print(review, "→", sentiment)
```
